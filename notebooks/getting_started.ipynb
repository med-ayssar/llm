Download [the-verdict.txt](https://www.kaggle.com/datasets/rakibulhasanshaon69/the-verdict-txt) into `data/the-verdict-txt/`.

A public download works without a token. If Kaggle asks you to sign in, put an API token in `~/.kaggle/access_token` or a legacy `~/.kaggle/kaggle.json` from [kaggle.com/settings/api](https://www.kaggle.com/settings/api).

```python
from pathlib import Path

import kagglehub

project_root = Path.cwd()
if not (project_root / "pyproject.toml").exists():
    project_root = project_root.parent

output_dir = project_root / "data" / "the-verdict-txt"
output_dir.mkdir(parents=True, exist_ok=True)

dataset_path = kagglehub.dataset_download(
    "rakibulhasanshaon69/the-verdict-txt",
    output_dir=str(output_dir),
)

verdict = Path(dataset_path) / "the-verdict.txt"
print(verdict)
print(f"{verdict.stat().st_size} bytes")
```
