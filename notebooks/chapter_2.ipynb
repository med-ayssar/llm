```python
from llm import data, tokenizer
import re

store = data.Data()
```

```python
items = store.list(extension="txt")
item = items[0]
vocabulary = store.data(item.path)
```

```python
vocabulary
```

```python
tokens = re.split(r'([,;:?_!"()\\ \s]|--)', vocabulary)
tokens
```

```python
spliter = tokenizer.Spliter()
simpleTokenizer = tokenizer.SimpleTokenizer(vocab=vocabulary, spliter=spliter)
```

```python
simpleTokenizer
```

```python
text="""It's the last he painted, you know,"
        Mrs. Gisburn said with pardonable pride."""
ids = simpleTokenizer.encode(text)
ids
```

```python
print(simpleTokenizer.decode(ids))
```
